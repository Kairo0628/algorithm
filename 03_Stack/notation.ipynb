### **전위 표기법(Prefix)**

연산자가 피연산자 앞에 위치하는 방식
- 중위 표기법: A + B * C
- 전위 표기법: + A * B C

```python
from stack import Stack

operators = '+-*/('

prefix1 = '+*ABC'
prefix2 = '-+AB/CD'
prefix3 = '*+A/BC-D*EF'

for n in [prefix1, prefix2, prefix3]:
    s = Stack(100)

    for i in n:
        if i not in operators:
            y = i
            while not s.is_empty() and s.peek() not in operators:
                x = s.pop()
                op = s.pop()            
                y = (f'({x} {op} {y})')
            s.push(y)
        else:
            s.push(i)
    print(s.pop())
```

```python
prefix1 = '+*324'
prefix2 = '-+35/84'
prefix3 = '*+1/93-8*23'

operators = '+-*/'

for n in [prefix1, prefix2, prefix3]:
    s = Stack(100)
    rev_n = n[::-1]
    for i in rev_n:
        if i not in operators:
            s.push(i)
        else:
            x = s.pop()
            y = s.pop()
            ans = eval(f'{x} {i} {y}')
            s.push(ans)
    
    print(s.pop())
```

### **후위 표기법(Postfix)**

연산자가 피연산자 뒤에 위치하는 방식
- 중위 표기법: A + B * C
- 후위 표기법: A B C * +

```python
from stack import Stack

postfix1 = 'AB*C+'
postfix2 = 'AB+CD/-'
postfix3 = 'ABC/+DEF*-*'

operators = '+-*/'

for n in [postfix1, postfix2, postfix3]:
    s = Stack(100)
    
    for i in n:
        if i not in operators:
            s.push(i)
        else:
            y = s.pop()
            x = s.pop()
            s.push(f'({x} {i} {y})')
        
    print(s.pop())
```
