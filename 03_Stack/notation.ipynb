### **전위 표기법(Prefix)**

연산자가 피연산자 앞에 위치하는 방식
- 중위 표기법: A + B * C
- 전위 표기법: + A * B C

```python
operator_priority = {
    '+': 1,
    '-': 1,
    '/': 2,
    '*': 2
}

from stack import Stack

operators = '+-*/('

s = Stack(100)

prefix1 = '+*ABC'
prefix2 = '-+AB/CD'
prefix3 = '*+A/BC-D*EF'

for i in prefix3:
    if i not in operators:
        curr = i
        while not s.is_empty() and s.peek() not in operators:
            operand = s.pop()
            operator = s.pop()            
            curr = (f'({operand} {operator} {curr})')
        s.push(curr)
    else:
        s.push(i)
s.pop()
```

```python
prefix1 = '+*324'
prefix2 = '-+35/84'
prefix3 = '*+1/93-8*23'

s = Stack(100)
operators = '+-*/'

for n in [prefix1, prefix2, prefix3]:
    rev_n = n[::-1]
    for i in rev_n:
        if i not in operators:
            s.push(i)
        else:
            x = s.pop()
            y = s.pop()
            ans = eval(f'{x} {i} {y}')
            s.push(ans)
    
    print(s.pop())
```

### **후위 표기법(Postfix)**

연산자가 피연산자 뒤에 위치하는 방식
- 중위 표기법: A + B * C
- 후위 표기법: A B C * +
