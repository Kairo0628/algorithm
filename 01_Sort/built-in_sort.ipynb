모든 파이썬 내장 정렬은 **Timsort** 방식을 따른다.

Timsort는 Merge Sort(병합 정렬)과 Insertion Sort(삽입 정렬)을 혼합하여 작동하는 알고리즘이다.

### list.sort()
- in-place 방식
- $O(n log n)$

```python
x = [2, 5, 4, 1, 3]

x.sort()
x
```

### sorted()
- 모든 iterable을 정렬
- $O(n log n)$

```python
x = [2, 5, 4, 1, 3]

sorted(x) # sorted()의 반환값이 존재. (반환값) != (입력 배열)
```

### key parameter

```python
x = [(4, 'A'), (3, 'A'), (1, 'B'), (5, 'B'), (2, 'B')]

x.sort(key = lambda x: x[0]) # 숫자 기준 정렬(0번 아이템 값 기준 정렬)
x
```

```python
x = [(4, 'A'), (3, 'A'), (1, 'B'), (5, 'B'), (2, 'B')]

x.sort(key = lambda x: x[1]) # 알파벳 기준 정렬(1번 아이템 값 기준 정렬)
x
```

다중 정렬 기준 적용
- 파이썬의 내장 정렬 함수는 모두 Stable Sort이다. 즉, 입력 순서가 보장된다. 따라서 원하는 순서로 정렬하기 위해서는 여러 조건을 지정해야 한다.
- 첫 번째 정렬 기준은 0번 아이템(숫자)을 기준으로 한다. 따라서 1번 아이템이 B인 경우가 전부 앞에 있으므로 B가 먼저 등장한다.
- 두 번째 정렬 기준은 0번 아이템을 기준으로 정렬한 뒤, 1번 아이템을 기준으로 정렬한다. 따라서 A가 먼저 등장한다.

```python
x = [(1, 'B'), (3, 'B'), (2, 'B'), (1, 'A'), (3, 'A'), (2, 'A')]

x.sort(key = lambda x: x[0])
x
```

```python
x = [(1, 'B'), (3, 'B'), (2, 'B'), (1, 'A'), (3, 'A'), (2, 'A')]

x.sort(key = lambda x: [x[0], x[1]])
x
```

### 내림차순 정렬

reverse 파라미터를 이용하여 내림차순 정렬을 할 수 있다.

```python
x = [2, 5, 4, 1, 3]

x.sort(reverse = True)
x
```
